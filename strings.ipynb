Accessing substring from a string

```python
# Concept of Indexing
s = "Hello"
print(s[0])  # H
print(s[1])  # e
```

```python
# Types of Indexing:
# 1. Positive Indexing
print(s[0])  # H
print(s[1])  # e

# 2. Negative Indexing
print(s[-1])  # o
print(s[-2])  # l
```

Slicing

```python
c = "Hello World"

print(c[0:5])  # Hello
print(c[6:])   # World
```

```python
print(c[2:])
```

```python
print(c[:5])   # Hello
```

```python
print(c[:])
```

```python
print(c[2:7:2])
```

```python
print(c[-5:-1:2])
```

```python
print(c[::-1])  # Reverse the string
```
