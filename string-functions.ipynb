Common Functions :

- len

- max

- min

- sorted


```python
c = "Kolkata"

len(c)  # 7
```

```python
max(c)  # t
```

```python
min(c)  # K
```

```python
sorted(c)
```

```python
sorted(c, reverse=True)  # ['t', 'o', 'l', 'k', 'a', 'a', 'K'] -> decending order
```

1. Capitalize/Title/Upper/Lower/Swapcase

```python
"It is raining today".capitalize()  # make the first letter capital and rest small
```

```python
"It is raining today".title()  # make the first letter of each word capital and rest small
```

```python
c.upper().lower()  # kolkata
```

```python
c.lower()  # kolkata
```

```python
"KolkAtA".swapcase()  # kOLKATA
```

2. Count 

```python
"It is raining today".count("a")
```

3. Find/Index

```python
"It is raining".find("is")
```

4. endswith/startswith

```python
"It is raining today".endswith("ing")  # False
```

```python
"It is raining today".startswith("It")  # True
```

5. Format

```python
"Hello my name is {} and I am {}".format("Yashraj", 21)
```

```python
"Hello my name is {1} and I am {0}".format("Yashraj", 21)
```

```python
name = input("Enter your name: ")
age = input("Enter your age: ")

"Hello my name is {name} and I am {age}".format(name=name, age=age)
```
