```python
def check_for_word(word):
    data = True
    with open("practice.txt", "r") as f:
        while data:
            data = f.read()
            if word in data:
                print("Word found")
            else:
                print("Word not found")
        
check_for_word(word = "xlearning")
```

```python
def check_for_word(word):
    data = True
    line_no = 1
    with open("practice.txt", "r") as f:
        while data:
            data = f.readline()
            if word in data:
                print(line_no)
                print("Word found")
                break
            line_no += 1
        else:
            print("Word not found")
        
check_for_word(word = "learning")
```

```python
with open("number.txt", "r") as f:
    data = f.read()
    print(data)
    
    num= ""
    for i in range(len(data)):
        if (data[i] == ","):
            print(int(num))
            num = ""
        else:
            num += data[i]
```

```python
count = 0
with open("number.txt", "r") as f:
    data = f.read()
    
    nums = data.split(",")
    for val in nums:
        if(int(val) % 2 == 0):
            count += 1
print(count)
```
